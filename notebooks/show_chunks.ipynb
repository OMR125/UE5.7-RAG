```python
import json
from pathlib import Path

import pandas as pd
from IPython.display import display

ROOT = Path.cwd().parent                      # the notebook lives in notebooks/
CHUNKS = ROOT / "data" / "parsed" / "chunks.jsonl"
print(CHUNKS, CHUNKS.exists())                # should print True

chunks = [json.loads(line) for line in CHUNKS.open(encoding="utf-8")]

df = pd.DataFrame(chunks)
df["pages"] = df["pages"].apply(lambda p: f"{p[0]}-{p[-1]}" if len(p) > 1 else str(p[0]))
df["chars"] = df["embed_text"].str.len()
pd.set_option("display.max_colwidth", None)

display(df[["id", "type", "pages", "chars", "headings"]].head(20))
```

```python
# a random sample of each type
display(df[df.type == "text"].sample(5, random_state=1)[["pages", "embed_text"]])
display(df[df.type == "table"].sample(3, random_state=1)[["pages", "embed_text", "return_text"]])

# everything touching one page
display(df[df.pages.str.contains("327")][["id", "type", "embed_text"]])

# the biggest table chunks
display(df[df.type == "table"].nlargest(5, "chars")[["id", "pages", "chars", "embed_text"]])

# text chunks that are mostly heading
display(df[(df.type == "text") & (df.chars - df.headings.str.len() < 40)][["id", "embed_text"]])
```
