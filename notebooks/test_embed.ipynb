```python
import chromadb
from pathlib import Path
from sentence_transformers import SentenceTransformer

ROOT = Path.cwd().parent
model = SentenceTransformer("BAAI/bge-small-en-v1.5", device="cuda")
collection = chromadb.PersistentClient(path=str(ROOT / "chroma_db")).get_collection("unreal_blueprints")

def search(query, n=5, where=None):
    vector = model.encode([query], normalize_embeddings=True).tolist()
    res = collection.query(query_embeddings=vector, n_results=n, where=where)
    for id_, dist, meta, doc in zip(res["ids"][0], res["distances"][0], res["metadatas"][0], res["documents"][0]):
        print(f'{id_} | dist {dist:.3f} | pages {meta["pages"]} | {meta["headings"]}')
        print("   ", doc[:200].replace("\n", " "))

```

```python
search("How does the FlipFlop node alternate its outputs?")
```

```python
search("What does the Loop checkbox do on a Timeline?")
```

```python
search("Get Play Rate", where={"type": "table"})
```
